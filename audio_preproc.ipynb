```python
def extract_features(input_path, sr=16000, n_mfcc=13):
    try:
        y, sr = librosa.load(input_path, sr=sr)
        
        pitch = librosa.core.piptrack(S=librosa.stft(y), sr=sr)[0].mean()
        rms = librosa.feature.rms(y=y).mean()
        mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=n_mfcc)
        mel_spec = librosa.feature.melspectrogram(y=y, sr=sr)
        chroma = librosa.feature.chroma_stft(y=y, sr=sr)
        zcr = librosa.feature.zero_crossing_rate(y)
        spectral_contrast = librosa.feature.spectral_contrast(y=y, sr=sr)
        
        features = {
            "file_name": os.path.basename(input_path),
            "pitch_mean": pitch,
            "rms_mean": rms,
            "zcr_mean": np.mean(zcr),
            "zcr_std": np.std(zcr),
        }
        
        # Flatten multi-value features like MFCC, MelSpec etc.
        for i, val in enumerate(np.mean(mfcc, axis=1)):
            features[f"mfcc_mean_{i+1}"] = val
        for i, val in enumerate(np.std(mfcc, axis=1)):
            features[f"mfcc_std_{i+1}"] = val
        for i, val in enumerate(np.mean(mel_spec, axis=1)):
            features[f"mel_spec_mean_{i+1}"] = val
        for i, val in enumerate(np.std(mel_spec, axis=1)):
            features[f"mel_spec_std_{i+1}"] = val
        for i, val in enumerate(np.mean(chroma, axis=1)):
            features[f"chroma_mean_{i+1}"] = val
        for i, val in enumerate(np.std(chroma, axis=1)):
            features[f"chroma_std_{i+1}"] = val
        for i, val in enumerate(np.mean(spectral_contrast, axis=1)):
            features[f"spectral_contrast_mean_{i+1}"] = val
        for i, val in enumerate(np.std(spectral_contrast, axis=1)):
            features[f"spectral_contrast_std_{i+1}"] = val
        
        return features
    except Exception as e:
        print(f"Error extracting from {input_path} : {e}")
        return None
```

```python
input_folder = r'C:\Users\Rishi S Etagi\Desktop\deep learning\torgo'  
```

```python
# list to store all data
data = []

# find all flac files
flac_files = [os.path.join(input_folder, f) for f in os.listdir(input_folder) if f.endswith('.wav')]

print(f"Found {len(flac_files)} .wav files")

for file in tqdm(flac_files):
    features = extract_features(file)
    if features:
        features['file_name'] = os.path.basename(file)
        data.append(features)

df = pd.DataFrame(data)

# save to csv
df.to_csv("torgo_features.csv", index=False)

print("Feature extraction complete. Saved to audio_features.csv")
```
